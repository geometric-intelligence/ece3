# Frequency

$\color{#EF5645}{\text{Definition}}$: The **frequency** of a category is the number of times it appears in a dataset.

$\color{#003660}{\text{In Python}}$, we use the function `value_counts()` from `pandas`. 

**Example:**  
Consider survey responses for favorite colors:  
`['Red', 'Blue', 'Red', 'Green', 'Blue', 'Red']`

**Insight** You can get mode easily once you know frequency!

$\color{#047C91}{\text{Exercise}}$ How would we find the frequencies without a built in function?

```python
def count_frequencies(items):
    freq = {}
    for x in items:
        if x in freq:
            freq[x] += 1
        else:
            freq[x] = 1
    return freq

data = ['A', 'B', 'A', 'C', 'A', 'B']
print(count_frequencies(data))
```

## 📊 `Series.value_counts()` — Parameters Explained

**Function:**  
`Series.value_counts(normalize=False, sort=True, ascending=False, bins=None, dropna=True)`

---

### 🔹 **Parameters**

| Parameter | Type | Default | Meaning |
|------------|------|----------|----------|
| **normalize** | `bool` | `False` | If `True`, returns **relative frequencies** (proportions) instead of raw counts. <br>• `False` → counts (e.g., 50) <br>• `True` → fractions (e.g., 0.50) |
| **sort** | `bool` | `True` | Sorts the output by **frequency** (most common first). <br>Set to `False` to preserve the original order of appearance. |
| **ascending** | `bool` | `False` | If `True`, sorts in **ascending order** (least frequent first). <br>Used only when `sort=True`. |
| **bins** | `int` | `None` | For **numeric data only** — splits values into equal-width **bins** (intervals), like `pd.cut`. <br>Example: `bins=4` groups ages into 4 ranges. |
| **dropna** | `bool` | `True` | Whether to **ignore missing values** (`NaN`). <br>Set to `False` to include them in the result. |

---


```python
import pandas as pd

colors = pd.Series(['Red', 'Blue', 'Red', 'Green', 'Blue', 'Red', 'Blue', 'Blue'])
colors.value_counts()
```

```python
colors.value_counts(sort=False)
```

```python

colors.value_counts(ascending=True)
```

$\color{#047C91}{\text{Exercise}}$: When would frequency be a useful metric over relative frequency and vice versa?

Hint: You've already seen both this class.

At a local animal shelter, there are 100 dogs and 2 cats.
One day, one dog is adopted together with one of the cats.

A volunteer posts on the notice board:

“1% of the dogs got adopted with 50% of the cats!”

Use frequency when the absolute number matters — when you care about how many observations there are, not just how they compare proportionally.

Ex:
**Scenario:**  
A hospital wants to know how many patients came in with **pneumonia** last month.

---

- **Frequency** gives the **actual number of cases** — e.g., **127 admissions**.  
- This information is critical for:
  - 🧑‍⚕️ Resource planning  
  - 👩‍💼 Staff scheduling  
  - 💰 Budgeting and supply management  
- A **percentage alone** (e.g., *5% of all visits*) hides the real workload —  the hospital needs *counts* to make operational decisions.

**Scenario:**  
Two hospitals report their number of pneumonia cases last month:

| Hospital | Pneumonia Cases | Total Patients | Percent |
|-----------|----------------:|----------------:|---------:|
| City General | 127 | 2,540 | **5%** |
| Valley Regional | 90 | 600 | **15%** |

---

- **Frequencies** (127 vs. 90) make it look like City General has a bigger problem.  
- But **percentages** reveal that pneumonia was **three times more common** at Valley Regional.  
- **Percentages** let us **compare groups of different sizes** fairly.

# Continuous Data: Histogram


$\color{#EF5645}{\text{Definition}}$ A **histogram** is a visual representation of the distribution of quantitative data. It is a diagram consisting of rectangles whose area is proportional to the frequency of a variable and whose width is equal to the class interval.

$\color{#003660}{\text{In Python}}$, we use the function `hist()` from `matplotlib.pyplot`. The default is 10 bins.


# 📈 Histogram: What It Shows

Histograms reveal the **shape** of your data — the pattern that numbers alone can’t show.

---

## 🔹 Common Distribution Shapes

| Shape | Description | Example |
|-------|--------------|----------|
| 🟢 **Normal** *(Bell Curve)* | Most values cluster around the mean; symmetrical shape | Human heights, test scores |
| 🟠 **Right-Skewed** *(Positive Skew)* | Long tail on the **right**; many low/moderate values, few high ones | Income, house prices |
| 🔵 **Left-Skewed** *(Negative Skew)* | Long tail on the **left**; many high values, few low ones | Age at retirement, grades in easy classes |

# 📦 Continuous Data: Box Plot

$\color{#EF5645}{\text{Definition:}}$  
A **box plot** (or **box-and-whisker plot**) is a visual summary of how data are **spread** and **centered**.  
It divides the dataset into **quartiles** — showing the **median**, **25th (Q1)**, and **75th (Q3)** percentiles — and highlights **outliers** beyond the typical range.

---

$\color{#00A2B8}{\text{Connection to Percentiles from last class:}}$  
- Percentiles divide ordered data into **100 equal parts**.  
- **Quartiles** divide data into **four equal parts**:  
  - **Q1 (25th percentile):** 25% of data fall below this value  
  - **Q2 (50th percentile / Median):** half of data fall below this value  
  - **Q3 (75th percentile):** 75% of data fall below this value  
- The **Interquartile Range (IQR)** = Q3 − Q1 → measures the **middle 50%** of your data.

---

<div style="text-align:center; margin-top:30px;">
  <img src="figs/03_boxplot.webp" alt="Time to process images vs sentences" width="1300px">
  <p style="font-size:14px



### Steps

1. Compute $ Q_1 $ and $ Q_3 $ — the 25th and 75th percentiles.  
2. Compute the interquartile range (IQR):

$$
\text{IQR} = Q_3 - Q_1
$$

3. Define whisker limits:

$$
\text{Lower bound} = Q_1 - 1.5 \times \text{IQR}
$$

$$
\text{Upper bound} = Q_3 + 1.5 \times \text{IQR}
$$

4. The **minimum** and **maximum** shown on the whiskers are the smallest and largest data points *within* these bounds.  
5. Any data points **beyond** those limits are plotted individually as **outliers**.

# Finding Q₁ and Q₃ Using the Median

When data are ordered, the **median (Q₂)** divides the set in half. There are two ways to form those halves when $n$ is odd.

**Example:**  
$X = [1,\; 2,\; 3,\; 4,\; 5,\; 6,\; 7]$  
Here, $n = 7$ and the median $Q_2 = 4$.

---

### 1. Exclusive Method
Exclude the median when splitting the data.  
Lower half: $[1,\,2,\,3]$ Upper half: $[5,\,6,\,7]$  
$Q_1 = 2,$ $Q_3 = 6$  
Used in many statistics textbooks and AP Statistics.

---

### 2. Inclusive Method
Include the median in both halves.  
Lower half: $[1,\,2,\,3,\,4]$ Upper half: $[4,\,5,\,6,\,7]$  
$Q_1 = 2.5,$ $Q_3 = 5.5$  
Used by NumPy, pandas, Excel, and `matplotlib.boxplot()`.

# Categorical Data: Bar Chart


$\color{#EF5645}{\text{Definition:}}$  
A **bar chart** is a visual representation of the **frequency or proportion** of **categorical data**.  
Each bar represents a category, and the **height (or length)** of the bar is proportional to its count or percentage.  
Unlike histograms, **bar charts display discrete categories**, not continuous intervals.

---

$\color{#003660}{\text{In Python:}}$  
We use the function `bar()` or `value_counts().plot(kind='bar')` from `matplotlib.pyplot` or `pandas`.

---

# Categorical Data: Pie Chart

$\color{#EF5645}{\text{Definition}}$  
A **pie chart** shows how a categorical variable divides a whole into parts. Each slice represents a category’s **proportion or percentage** of the total.

Pie charts are most useful when there are only a few categories and you want to emphasize **relative share**.

---

$\color{#003660}{\text{In Python:}}$  
We can use the function `pie()` from `matplotlib.pyplot`.


```python
import matplotlib.pyplot as plt

# Example data
sizes = [40, 25, 20, 15]
labels = ['Apples', 'Bananas', 'Oranges', 'Berries']

plt.pie(sizes, labels=labels, autopct='%1.1f%%', startangle=90)
plt.title("Fruit Sales Distribution")
plt.show()
```

# The Pie Chart Controversy: When (Not) to Use It

**The uncomfortable truth:** Pie charts are often misused. They aren’t “bad,” they’re **specific**.

The pie chart fails when slices are too similar; angle/area judgments are poor and ranking is unclear

```python
companies = ['Co A','Co B','Co C','Co D','Co E','Co F','Co G','Other']
shares = [15, 14, 13, 12, 11, 10, 9, 16]

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 5), layout="constrained")

ax1.pie(shares, labels=companies, autopct='%1.1f%%', startangle=90)
ax1.set_title('Market Share (Pie): Hard to Compare')

ax2.barh(companies, shares, color='steelblue', edgecolor='black')
ax2.set_xlabel('Market Share (%)')
ax2.set_title('Market Share (Bar): Easy to Compare')
ax2.grid(axis='x', alpha=0.3)

plt.show()
```

**Good use case:** Few categories, one clearly dominant slice, message is “part of a whole.”

✅ **Use when:**  
- You have **2–5 categories**  
- One category is clearly **dominant (>50%)**  
- The goal is a **part-of-whole message**  
- You want **visual or emotional impact**, not precision

```python
categories = ['Programs','Operations','Fundraising','Admin']
percentages = [70, 15, 10, 5]
explode = (0.1, 0, 0, 0)   # emphasize the main slice

plt.figure(figsize=(7, 5))
plt.pie(percentages, labels=categories, autopct='%1.0f%%',
        startangle=90, explode=explode, textprops={'fontsize': 12})
plt.title('Where Your Donation Goes')
plt.show()
```
