## Part 5: Multiple Operations

### Exercise 5.1
Create a second dictionary called `prices` with:
- "apples": 2.5
- "bananas": 1.8
- "oranges": 3.0
- "grapes": 4.5

Print the prices dictionary.

````python
# Solution
prices = {"apples": 2.5, "bananas": 1.8, "oranges": 3.0, "grapes": 4.5}
print(prices)
````

### Exercise 5.2
Calculate the total value of all apples in inventory:
- Multiply the number of apples by the price per apple
- Store this in a variable called `apple_value`
- Print the result

````python
# Solution
apple_value = market_inventory["apples"] * prices["apples"]
print(apple_value)
````

### Exercise 5.3
Calculate the total value of bananas and oranges combined.
Print the result.

````python
# Solution
banana_value = market_inventory["bananas"] * prices["bananas"]
orange_value = market_inventory["oranges"] * prices["oranges"]
total_value = banana_value + orange_value
print(total_value)
````

## Bonus Challenge

Create a dictionary called `customer_order` with:
- "apples": 10
- "bananas": 5

Calculate and print:
1. The remaining apples in inventory after this order
2. The remaining bananas in inventory after this order
3. The total cost of this customer's order

````python
# Solution
customer_order = {"apples": 10, "bananas": 5}

# 1. Remaining apples
remaining_apples = market_inventory["apples"] - customer_order["apples"]
print("Remaining apples:", remaining_apples)

# 2. Remaining bananas
remaining_bananas = market_inventory["bananas"] - customer_order["bananas"]
print("Remaining bananas:", remaining_bananas)

# 3. Total cost
apple_cost = customer_order["apples"] * prices["apples"]
banana_cost = customer_order["bananas"] * prices["bananas"]
total_cost = apple_cost + banana_cost
print("Total cost:", total_cost)
````
