
# Summarizing Data with the Iris Dataset



**Goal:** Practice core data summarization and visualization skills on the Iris dataset. This is a commonly used dataset which you will come accross if you continue to take any data science or machine learning classes. 

The Iris dataset was used in R.A. Fisher's classic 1936 paper, [The Use of Multiple Measurements in Taxonomic Problems](https://rcs.chemometrics.ru/Tutorials/classification/Fisher.pdf). 

In irises, the petal (blade) and sepal (tepals) size—is tied to pollination and reproductive isolation. Petals primarily attract pollinators, while sepals protect buds and help present nectar guides; their sizes often co-vary with pollinator type, resource allocation, and developmental constraints. In this dataset, we look at three different species: Iris setosa, I. versicolor, and I. virginica. Plotting distributions (histograms/boxplots) shows how much traits vary within a species versus between species. Scatter plots reveal how different traits vary with eachother: i.e. does larger petal width mean shorter sepal length?
